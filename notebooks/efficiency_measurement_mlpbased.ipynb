# Putting it all together

All efficiency measurements are then copied to the `efficiency_eval_files` folder for further analysis and visualization.

```python
import pandas as pd


def compute_avg_time_per_tx(csv_path: str) -> float:
    df = pd.read_csv(csv_path)

    total_transactions = df["num_transactions"].sum()
    total_compute_seconds = df["compute_seconds"].sum()
    avg_seconds_per_tx = total_compute_seconds / total_transactions

    print(f"Total transactions processed: {total_transactions}")
    print(f"Total computation time: {total_compute_seconds:.6f} seconds")
    print(f"Average time per transaction: {avg_seconds_per_tx:.6f} seconds")

    return avg_seconds_per_tx


avg_xchaindatagen = compute_avg_time_per_tx("efficiency_eval_files/graph_generation_timing_xchaindatagen.csv")
```

```python
import math


def compute_avg_std_per_tx(csv_path: str) -> tuple[float, float]:
    df = pd.read_csv(csv_path)

    avg_per_graph = df["average_per_graph"].iloc[0]
    std_per_graph = math.sqrt(df["var_per_graph"].iloc[0]) # std = sqrt(variance)

    print(f"Average time per tx: {avg_per_graph:.6f} seconds")
    print(f"Standard deviation per tx: {std_per_graph:.6f} seconds")

    return avg_per_graph, std_per_graph


print("Single-chain dataset:")
avg_single, std_single = compute_avg_std_per_tx("efficiency_eval_files/pyg_dataset_single.csv")

print("\nCross-chain dataset:")
avg_cctx, std_cctx = compute_avg_std_per_tx("efficiency_eval_files/pyg_dataset_cctx.csv")
```

```python
def compute_stage_mean_std(csv_path: str) -> pd.DataFrame:
    df = pd.read_csv(csv_path)
    stage_stats = df.set_index("stage")[["mean_ms_per_tx", "std_ms_per_tx"]]

    for stage, row in stage_stats.iterrows():
        print(f"{stage}: mean={row['mean_ms_per_tx']:.6f} ms/tx, std={row['std_ms_per_tx']:.6f} ms/tx")

    return stage_stats


timing_report_files = {
    ("deepsad", "single"): "efficiency_eval_files/timing_report_deepsad_single.csv",
    ("deepsad", "cctx"): "efficiency_eval_files/timing_report_deepsad_cctx.csv",
    ("mlpbased", "single"): "efficiency_eval_files/timing_report_mlpbased_single.csv",
    ("mlpbased", "cctx"): "efficiency_eval_files/timing_report_mlpbased_cctx.csv",
}

timing_report_stats = {}
for (model, scope), path in timing_report_files.items():
    print(f"\n{model} ({scope}):")
    timing_report_stats[(model, scope)] = compute_stage_mean_std(path)
```

```python
import numpy as np

STAGE_ROWS = [
    "Tx Graph Generation",
    "PyTorch Conversion",
    "Normalization",
    "Metapath Aggregation",
    "Inference (forward, batch=1)",
]

MODEL_SCOPE_COLUMNS = [
    ("mlpbased", "single", "\\makecell[c]{MLP-based\\\\(single-chain)}"),
    ("deepsad", "single", "\\makecell[c]{DeepSAD-based\\\\(single-chain)}"),
    ("mlpbased", "cctx", "\\makecell[c]{MLP-based\\\\(cross-chain)}"),
    ("deepsad", "cctx", "\\makecell[c]{DeepSAD-based\\\\(cross-chain)}"),
]

# Tx Graph Generation is shared across every model/scope combination
xchaindatagen_ms = avg_xchaindatagen * 1000

# PyTorch Conversion is shared across the two models within the same scope
pyg_ms_per_scope = {
    "single": compute_avg_std_per_tx("efficiency_eval_files/pyg_dataset_single.csv")[0] * 1000,
    "cctx": compute_avg_std_per_tx("efficiency_eval_files/pyg_dataset_cctx.csv")[0] * 1000,
}


def stage_ms(model: str, scope: str, stage: str) -> float:
    return timing_report_stats[(model, scope)].loc[stage, "mean_ms_per_tx"]


efficiency_table = pd.DataFrame(
    index=STAGE_ROWS,
    columns=[label for _, _, label in MODEL_SCOPE_COLUMNS],
    dtype=float,
)

for model, scope, label in MODEL_SCOPE_COLUMNS:
    efficiency_table.loc["Tx Graph Generation", label] = xchaindatagen_ms
    efficiency_table.loc["PyTorch Conversion", label] = pyg_ms_per_scope[scope]
    efficiency_table.loc["Normalization", label] = stage_ms(model, scope, "normalization")
    efficiency_table.loc["Metapath Aggregation", label] = stage_ms(model, scope, "metapath_aggregation")
    efficiency_table.loc["Inference (forward, batch=1)", label] = stage_ms(model, scope, "forward_batch1")

efficiency_table.loc["Total"] = efficiency_table.loc[STAGE_ROWS].sum()
efficiency_table.loc["TPS"] = 1 / (efficiency_table.loc["Total"] / 1000)

efficiency_table
```

```python
def dataframe_to_latex(
    df: pd.DataFrame,
    float_fmt: str = "{:.3f}",
    caption: str = "Per-transaction processing time (ms) by stage, model and scope.",
    label: str = "table:evaluation-efficiency-stages",
) -> str:
    n_cols = len(df.columns)
    lines = [
        "\\begin{table}",
        "%\\centering",
        "\\resizebox{\\linewidth}{!}{%",
        "\\begin{tabular}{l" + "c" * n_cols + "}",
        "\\toprule",
        "Stage & " + " & ".join(df.columns) + " \\\\",
        "\\midrule",
    ]

    for stage, row in df.iterrows():
        values = row.to_numpy(dtype=float)
        cells = [float_fmt.format(value) for value in values]

        if stage == "Total":
            lines.append("\\midrule")
        lines.append(f"{stage} & " + " & ".join(cells) + " \\\\")

    lines.append("\\bottomrule")
    lines.append("\\end{tabular}")
    lines.append("}")
    lines.append(f"\\caption{{{caption}}}")
    lines.append(f"\\label{{{label}}}")
    lines.append("\\end{table}")
    return "\n".join(lines)


print(dataframe_to_latex(efficiency_table))
```
