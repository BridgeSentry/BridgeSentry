````python
import numpy as np
import torch

# check HeteroData graphs for NaNs in node and edge matrices

for i, graph in enumerate(internaltx_dataset):
    tx_hash = getattr(graph, "cctx_id", None) or getattr(graph, "tx_hash", None) or getattr(graph, "id", None) or i

    # nodes
    for ntype in graph.node_types:
        for attr in graph[ntype].keys():
            val = graph[ntype][attr]
            if isinstance(val, torch.Tensor):
                arr = val.detach().cpu().numpy()
                if np.issubdtype(arr.dtype, np.floating) and np.isnan(arr).any():
                    print(f"Graph with cctx_id {tx_hash} has NaN in node_type {ntype} matrix '{attr}'")
                    print(val)
            elif isinstance(val, np.ndarray):
                if np.issubdtype(val.dtype, np.floating) and np.isnan(val).any():
                    print(f"Graph with cctx_id {tx_hash} has NaN in node_type {ntype} matrix '{attr}'")
                    print(val)

    # edges
    for etype in graph.edge_types:
        for attr in graph[etype].keys():
            val = graph[etype][attr]
            if isinstance(val, torch.Tensor):
                arr = val.detach().cpu().numpy()
                if np.issubdtype(arr.dtype, np.floating) and np.isnan(arr).any():
                    print(f"Graph with cctx_id {tx_hash} has NaN in edge_type {etype} matrix '{attr}'")
                    print(val)
            elif isinstance(val, np.ndarray):
                if np.issubdtype(val.dtype, np.floating) and np.isnan(val).any():
                    print(f"Graph with cctx_id {tx_hash} has NaN in edge_type {etype} matrix '{attr}'")
                    print(val)
````

# Comparing Single Chain Dataset Changes

````python
# Load the current internal tx's dataset
import sys

sys.path.insert(0, '..')

import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
from tqdm import tqdm

from dataset_generator.single_transactions_dataset import SingleTransactionsDataset
from dataset_generator.feature_extraction import (
    NODE_TYPE_FEATURE_DIMS, SINGLE_CHAIN_LABEL_MAP,
    IN_DEGREE_INDEX, OUT_DEGREE_INDEX, AMOUNTS_INDEX,
)

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)

internaltx_dataset = SingleTransactionsDataset(root="../datasets/internaltx-single-all-until-thorchain-except-omni")
print(f"Dataset size: {len(internaltx_dataset)} graphs")
````

````python
# Load the older event-driven dataset
eventdriven_dataset = SingleTransactionsDataset(root="../datasets/single-allauto-harmony-polynetwork-thorchain")
print(f"Dataset size: {len(eventdriven_dataset)} graphs")
````

For each graph in the new dataset, check there is a corresponding graph with the same tx_hash in the older dataset.

 Place each correspondence in an indice map like so

 `tx_hash -> (old_index, new_index)`
 Place in a dictionary all graphs that are missing from either dataset, with the dict structured as follows:
 ```missing_graphs = {
 "bridge_name_1": [
   (tx_hash, "OLD"/"NEW", index in dataset),
   ...   
],
 "bridge_name_2": [
   (tx_hash, "OLD"/"NEW", index in dataset),
   ...
 ]
 ...
}
```

````python
# Build a mapping of tx_hash to (index, bridge_name) for both datasets
new_index_map = {}
old_index_map = {}

for idx, graph in enumerate(internaltx_dataset):
    tx_hash = graph.tx_hash
    bridge = graph.bridge
    if tx_hash not in new_index_map:
        new_index_map[tx_hash] = []
    new_index_map[tx_hash].append((idx, bridge))

for idx, graph in enumerate(eventdriven_dataset):
    tx_hash = graph.tx_hash
    bridge = graph.bridge
    if tx_hash not in old_index_map:
        old_index_map[tx_hash] = []
    old_index_map[tx_hash].append((idx, bridge))
````

````python
# Find correspondences and missing graphs
correspondence_map = {}
missing_graphs = defaultdict(list)

# Check all tx_hashes in new dataset
for tx_hash, new_entries in new_index_map.items():
    if tx_hash in old_index_map:
        for new_idx, bridge in new_entries:
            correspondence_map[tx_hash] = (old_index_map[tx_hash][0][0], new_idx)
    else:
        for new_idx, bridge in new_entries:
            missing_graphs[bridge].append((tx_hash, "NEW", new_idx))

# Check for tx_hashes only in old dataset
for tx_hash, old_entries in old_index_map.items():
    if tx_hash not in new_index_map:
        for old_idx, bridge in old_entries:
            missing_graphs[bridge].append((tx_hash, "OLD", old_idx))
````

````python
print(f"Found {len(correspondence_map)} corresponding graphs")
for bridge, txs in missing_graphs.items():
    count_old = sum(1 for tx in txs if tx[1] == "OLD")
    count_new = sum(1 for tx in txs if tx[1] == "NEW")
    print(f"Bridge {bridge}: {count_old} missing in new dataset, {count_new} missing in old dataset")
````
